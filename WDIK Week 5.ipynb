```python
df['Customer_Name'] = df['Customer_Name'].fillna('Unknown')
```

```python
print(df)
```

```python
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].mean())
df['Price'] = df['Price'].fillna(df['Price'].median())
df['Delivery_Days']=df['Delivery_Days'].fillna(df['Delivery_Days'].mean())
df['Country']=df['Country'].fillna('Unknown')
df['Product']=df['Product'].fillna('Unknown')
print(df)
#I chose median for price because the price values have more outliers
```

```python
df.isnull().sum()
```

It was hard deciding on whether to use the mean values for the delivery days and quantity. I had to check the Interquartile range of the values to see whether there were outliers or not.
I didn't drop the rows with the missing values for country and product because 20% of the dataset was at risk. These columns were also not key identifiers.
