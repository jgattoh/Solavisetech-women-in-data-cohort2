```python
import pandas as pd
data = {
'Order_ID':   ['ORD001','ORD002','ORD003','ORD004','ORD005', 
'ORD006','ORD007','ORD008','ORD009','ORD010'], 
'Product':    ['Laptop','Rice Bag','Headphones','Notebook','Blender', 
'Rice Bag','Laptop','Notebook','Blender','Headphones'], 
'Category':   ['Electronics','Food','Electronics','Stationery','Appliances', 
'Food','Electronics','Stationery','Appliances','Electronics'], 
'Region':     ['North','South','East','West','North', 
'East','South','North','West','North'], 
'Units_Sold': [5, 30, 12, 50, 8, 20, 3, 80, 6, 15], 
'Unit_Price': [850.00, 12.50, 45.00, 3.20, 65.00,  
12.50, 850.00, 3.20, 65.00, 45.00], 
'Month':      ['January','January','February','February','March', 
'March','April','April','May','May'] 
}
df = pd.DataFrame(data) 
print(df)
```

SELECTING COLUMNS

```python
print(df[['Product','Region']]) #Showing only the Product and Region columns
```

```python
print(df[['Order_ID','Category','Units_Sold']])
```

```python
print(df['Units_Sold'])# This shows the datatype and displays the data as a series.
```

```python
print(df[['Units_Sold']])# This puts the data in a wider frame, and the data type is not shown.
```

FILTERING ROWS

```python
north = df[df["Region"] == "North" ]
print (north)
```

```python
Higher_Units_Sold = df[df["Units_Sold"] > 10 ]
print(Higher_Units_Sold)
```

```python
Electronics_5 = df[(df['Category'] == 'Electronics') & (df['Units_Sold']> 5)]
print(Electronics_5)
```

```python
South_West = df[(df['Region'] == 'South') | (df['Region'] == 'West')]
print(South_West)
```

GROUPING DATA

```python
#Finding the total number of units sold for each Region
```

```python
Total_Units_Sold = df.groupby('Region')['Units_Sold'].sum()
print(Total_Units_Sold)
```

```python
#Finding the average unit price for each category
```

```python
Average_Unit_Price = df.groupby('Category')['Unit_Price'].mean()
print(Average_Unit_Price)
```

Find the total Units_Sold AND the average Unit_Price for each Category in one step. 
Hint: Use .agg() 
 


```python
summary = df.groupby('Category').agg({
    'Units_Sold' : 'sum',
    'Unit_Price' : 'mean' })
print(summary)
```

Find the total Units_Sold for each Month

```python
Total_Units_Per_Month = df.groupby('Month')['Units_Sold'].sum()
print(Total_Units_Per_Month)
```

Which region had the highest total Units_Sold?

```python
Highest_Units = df.groupby('Region')['Units_Sold'].sum().idxmax()
print(Highest_Units)
```

ANSWERING BUSINESS QUESTION

Which product had the lowest average Unit_Price? 
Hint: Use groupby, then mean, then idxmin 


```python
Lowest_Price_Product = df.groupby('Product')['Unit_Price'].mean().idxmin()
print(Lowest_Price_Product)
```

Which month had the highest total Units_Sold? 

```python
Highest_Units_Month = df.groupby('Month')['Units_Sold'].sum().idxmax()
print(Highest_Units_Month)
```

Show only the Electronics orders from the North region. Display only the Product, Units_Sold and 
Unit_Price columns.

```python
Electronics_North = df[(df['Category'] == 'Electronics' ) & (df['Region'] == 'North') ] [['Product','Units_Sold','Unit_Price']]
print(Electronics_North)
```

Extra challenge: Create a new column called Total_Sales by multiplying Units_Sold by Unit_Price. Then find which 
category had the highest total sales overall. 

```python
df['Total_Sales'] = df['Units_Sold'] * df['Unit_Price']
Highest_Category = df.groupby('Category')['Total_Sales'].sum().idxmax()
print(Highest_Category)
```
